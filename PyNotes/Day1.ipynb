## 1. What is Python?

- **Definition:** Python is a high-level, interpreted, OOPs, general-purpose programming language designed for readable, concise/simple code.
- **Creator:** Guido van Rossum.
- **First public release:** 1991.
- **Name origin:** named after the comedy group *Monty Python*, not the snake.
- **Common uses:** web development, automation, scripting, data analysis, scientific computing, AI/ML, testing, and education.
- **Language family:** Python is a programming language; it is not an operating system, database, IDE, or framework.
- **Standard implementation:** CPython, which is written primarily in C.
- **File extension:** `.py` is the conventional extension for Python source files.
- **Notebook extension:** `.ipynb` is used by Jupyter Notebook files.
- **Case sensitivity:** Python is case-sensitive: `name`, `Name`, and `NAME` are three different identifiers.
- **Indentation:** indentation is syntactically meaningful; it defines suites/blocks instead of using braces like `{}`.
- **Comments:** `#` begins a comment outside a string.
- **Statement terminator:** a newline normally ends a statement; semicolons can separate simple statements on one line but are uncommon style.

## 2. Python's history — exam facts

- **Late 1980s:** Guido van Rossum began developing Python at CWI in the Netherlands.
- **1991:** first public release.
- **Python 2.0:** released in 2000; introduced important features including list comprehensions and garbage collection improvements.
- **Python 3.0:** released in 2008; intentionally introduced some backward-incompatible changes.
- **Python 2 end of life:** 1 January 2020; Python 2 should not be chosen for new projects.
- **Today:** use a supported Python 3 release for new learning and projects.
- **PEP:** Python Enhancement Proposal; a document proposing or describing changes, processes, or standards for Python.
- **PEP 8:** widely used style guide for Python code.
- **Zen of Python:** a set of design aphorisms viewable by running `import this`.

**MCQ trap:** Python 3 is not simply a version that can run every Python 2 program unchanged; some syntax and behavior differ.

## 4. Advantages and limitations of Python

### Advantages
- Readable syntax can make code easier to learn and maintain.
- Large standard library and third-party package ecosystem.
- Useful across automation, data, web, testing, and AI/ML tasks.
- Interactive use is supported by the REPL and Jupyter notebooks.
- Cross-platform implementations are available.
- Supports procedural, object-oriented, and functional approaches.
- Automatic memory management reduces routine manual memory handling.

### Limitations / trade-offs
- CPython can be slower than compiled languages such as C for some CPU-bound workloads.
- Dynamic typing can allow some type-related mistakes to appear at runtime rather than being caught before execution.
- The Global Interpreter Lock (GIL) in conventional CPython builds can limit parallel execution of Python bytecode across threads for CPU-bound work; details depend on Python version/build and workload.
- Packaging and dependency management can become complicated across projects.
- Mobile and low-level embedded development are not Python's strongest general-purpose areas compared with some alternatives.
- Memory use can be higher than in carefully optimized lower-level programs.

**MCQ trap:** “Python is always slow” and “Python cannot do multithreading” are both overgeneralizations.

## 6. Variables — definition and core idea

- **Variable (beginner definition):** a name used to refer to an object/value.
- **More precise Python model:** assignment binds a name to an object; it does not create a fixed, typed storage box in the way beginners sometimes imagine.
- A variable does not need a prior declaration in Python; a name is created/bound when assignment succeeds.
- Python is **dynamically typed**: the object has a type, and a name can later be rebound to an object of another type.
- Python is **not untyped**: objects have types, and Python checks operation compatibility at runtime.
- Multiple names can refer to the same object.
- Assignment does not necessarily copy an object.

```python
age = 25
name = "Ruchi"
name2 = """Ruchii"""
height = 1.52
is_student = True

print(age, name, name2, height, is_student)
print(type(age))
print(type(name))
print(type(name2))
print(type(height))
print(type(is_student))
```

## 7. Variable naming rules

### Valid rules
- A name may contain letters (`A–Z`, `a–z`), digits (`0–9`), and underscores (`_`).
- A name cannot begin with a digit.
- A name cannot contain spaces or punctuation such as `-`, `@`, or `!`.
- Names are case-sensitive.
- Python keywords cannot be used as ordinary identifiers.
- Unicode identifiers are allowed, but simple ASCII names are often more portable and readable.
- A leading underscore, such as `_count`, is a convention indicating an internal/non-public name; it is not a security mechanism.
- Names with two leading and two trailing underscores, such as `__init__`, are commonly special-method names defined by Python conventions.
- `__name` inside a class triggers name mangling; it is not the same as a special method name.

### Good style
- Use `snake_case` for variables and functions: `total_marks`.
- Use meaningful names: `student_count` is clearer than `sc`.
- Constants are conventionally uppercase: `MAX_RETRIES = 3`; this is a convention, not enforced immutability.

```python
# Valid identifiers
student_name = "Asha"
_marks2 = 88
total_marks = 450
name2 = "Mina"

# Invalid examples — uncomment one at a time to observe SyntaxError:
# 2name = "Mina"       # starts with a digit
# student-name = "A"   # hyphen is interpreted as subtraction
# first name = "A"     # spaces are not allowed
# class = "Python"     # 'class' is a keyword
```

## 8. Keywords, built-ins, and identifiers

- **Keyword:** a reserved word with a special syntactic role, e.g. `if`, `else`, `for`, `def`, `class`, `return`, `True`, `False`, `None`.
- **Identifier:** a name created by the programmer or used to refer to something, e.g. `total`, `calculate_sum`.
- **Built-in name:** a name supplied by Python, e.g. `print`, `len`, `sum`, `str`, `list`.
- Do not confuse keywords with built-ins: `if` is a keyword; `print` is a built-in function.
- Avoid naming a variable `list`, `str`, `sum`, or `print`, because it shadows the built-in name in the current scope.
- Check keywords with `keyword.kwlist`.
- Check whether a string is a valid identifier with `str.isidentifier()`. This alone does not prove it is not a keyword.

```python
import keyword

print(keyword.kwlist)
print("student_name".isidentifier())  # True
print("2student".isidentifier())      # False
print(keyword.iskeyword("for"))       # True
print(keyword.iskeyword("print"))     # False
```

## 9. Assignment operators and unpacking

- `=` assigns/binds a name; it does not mean “is equal to” as in mathematics.
- `==` tests equality.
- `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=` are augmented assignment operators.
- `a = b = 10` assigns the same object/value to both names.
- `a, b = 1, 2` performs iterable unpacking.
- `a, b = b, a` swaps bindings without a temporary variable.
- The number of unpacked targets must match the number of values unless starred unpacking is used.
- `first, *middle, last = [1, 2, 3, 4]` assigns `first=1`, `middle=[2, 3]`, `last=4`.
- Augmented assignment can behave differently from `x = x + y` for mutable objects because an in-place operation may modify an object.

```python
x = 10
x += 2
print(x)  # 12

a, b = 1, 2  # init 2 var in one line
print(a,b)
a, b = b, a  # swapping without temp
print(a, b)  # 2 1

ruchi, *kicchu, kajju = [1, 2, 3, 4]
print(ruchi, kicchu, kajju)  # 1 [2, 3] 4
```

## 10. Rebinding, identity, equality, and mutability

- **Rebinding:** assigning a name to a different object.
- **Equality (`==`):** asks whether two objects compare as equal.
- **Identity (`is`):** asks whether two references point to the exact same object.
- Use `==` for value comparison; use `is` especially for singleton checks such as `value is None`.
- **Mutable objects:** can be changed after creation; common examples are `list`, `dict`, and `set`.
- **Immutable objects:** cannot be changed in place; common examples are `int`, `float`, `bool`, `str`, and `tuple` (although a tuple can contain a mutable object).
- A list assignment such as `b = a` does not copy the list; both names refer to the same list.
- `b = a.copy()` creates a shallow copy for lists; nested mutable objects can still be shared.
- Small-integer and string interning/identity behavior can be implementation-dependent; do not use `is` to compare ordinary numbers or strings.

```python
a = [1, 2]
b = a
b.append(3)
print(a)       # [1, 2, 3] — same list object

c = a.copy()
c.append(4)
print(a)       # [1, 2, 3]
print(c)       # [1, 2, 3, 4]

p = [1, 2]
q = [1, 2]
print(p == q)  # True: equal contents
print(p is q)  # False: different list objects

result = None
print(result is None)  # True — preferred None check
```

## 11. Basic built-in data types

| Type | Example | Mutable? | Notes |
|---|---|---:|---|
| `int` | `42` | No | Integers have arbitrary precision, limited by available memory |
| `float` | `3.14` | No | Usually IEEE 754 double precision in CPython; many decimal fractions are approximations |
| `complex` | `2 + 3j` | No | `j` represents the imaginary part |
| `bool` | `True` | No | Only two values: `True` and `False`; `bool` is a subclass of `int` |
| `str` | `"hello"` | No | Unicode text; immutable |
| `list` | `[1, 2, 3]` | Yes | Ordered, mutable sequence |
| `tuple` | `(1, 2, 3)` | No | Ordered, immutable sequence container |
| `range` | `range(5)` | No | Represents an arithmetic progression; often used in loops |
| `dict` | `{"a": 1}` | Yes | Key-value mapping; keys must be hashable |
| `set` | `{1, 2, 3}` | Yes | Unique hashable elements; no positional indexing |
| `frozenset` | `frozenset({1, 2})` | No | Immutable set |
| `NoneType` | `None` | N/A | `None` represents absence/no value |

### Type-checking and conversion
- `type(value)` returns the object's type.
- `isinstance(value, SomeType)` is usually preferred when checking types because it accounts for subclass relationships.
- `int("12")` converts a valid integer string to an integer.
- `int("12.5")` raises `ValueError`; convert through `float` if truncation/conversion is intended.
- `float("3.5")` returns `3.5`.
- `str(value)` returns a string representation.
- `bool(value)` uses truth-value rules; empty collections, zero, `None`, and `False` are falsey, while most other objects are truthy.

```python
print(type(10))                  # <class 'int'>
print(isinstance(True, int))     # True
print(type(True) is int)         # False
print(int("12"))                # 12
# int("12.5")                   # ValueError
print(int(float("12.5")))        # 12 — truncates toward zero
print(bool([]), bool([0]))       # False True
```

## 13. Errors and exceptions — variable-related traps

- **SyntaxError:** code violates Python grammar; code may fail before the relevant statements run.
- **NameError:** a name is used but is not defined in the current accessible scope.
- **UnboundLocalError:** a subclass of `NameError`; a local variable is referenced before it has been assigned in a function.
- **TypeError:** an operation/function is used with an inappropriate type, e.g. `"2" + 3`.
- **ValueError:** type is acceptable but the value is unsuitable, e.g. `int("hello")`.
- **AttributeError:** an object does not have the requested attribute.
- **IndexError:** a sequence index is out of range.
- **KeyError:** a dictionary key is absent.
- **ZeroDivisionError:** division or modulo by zero for numeric operations where prohibited.
- **IndentationError:** indentation is invalid; it is a subclass of `SyntaxError`.
- **TabError:** inconsistent tabs and spaces; it is a subclass of `IndentationError`.

## 14. Scope: local, global, and nonlocal (intro)

- **Scope:** the region where a name can be accessed directly.
- **Local name:** assigned inside a function, unless declared otherwise.
- **Global name:** defined at module level.
- `global name` tells a function that assignments to that name should target the module-level binding.
- `nonlocal name` targets a binding in the nearest enclosing function scope, not the global scope.
- Reading a global name is different from assigning to it inside a function.
- Avoid unnecessary global state; passing values into functions and returning results is usually easier to test.

```python
count = 10

def show_count():
    count = 3  # local name; does not change the global count
    return count

print(show_count())  # 3
print(count)         # 10
```

## 15. Common MCQ facts and traps

1. Python is **case-sensitive**.
2. `=` is assignment; `==` is equality comparison; `is` is identity comparison.
3. `input()` returns a string, even when the user types digits.
4. `bool` is a subclass of `int`: `isinstance(True, int)` is `True`, but `type(True) is int` is `False`.
5. `True == 1` is `True`; `False == 0` is `True`; this does not mean their types are identical.
6. `None` is a singleton; test it with `is None` / `is not None`.
7. Empty list `[]`, empty tuple `()`, empty dictionary `{}`, empty set `set()`, empty string `""`, numeric zero, `False`, and `None` are falsey.
8. `{}` creates an empty dictionary; `set()` creates an empty set.
9. `{1, 1, 2}` becomes a set containing unique elements.
10. A one-item tuple needs a comma: `(5,)`; `(5)` is just the integer `5` in parentheses.
11. Lists are mutable; strings and tuples are immutable.
12. `list.sort()` sorts a list in place and returns `None`; `sorted(iterable)` returns a new sorted list.
13. `range(5)` produces values 0 through 4 when iterated; the stop value is excluded.
14. `and` and `or` short-circuit and return one of their operands, not necessarily a Boolean.
15. `not` returns a Boolean.
16. `==` may compare values; `is` compares object identity.
17. Python has no dedicated variable declaration syntax such as `int x = 5`.
18. A constant is conventionally uppercase; Python does not enforce constants.
19. `del name` removes a name binding; it does not guarantee immediate destruction of the object if other references exist.
20. A name can be rebound to an object of another type because Python is dynamically typed.
21. `type()` reports the immediate type; `isinstance()` also recognizes subclasses.
22. Python indentation is part of syntax.
23. Comments start with `#`; triple-quoted strings are string literals, not technically comments, although they are often used for docstrings.
24. `None`, `True`, and `False` are keywords/literals with special meanings.
25. `print` is a built-in function, not a keyword.
